## EDA

```python
# Shows the number of passengers who survived and did not survive.

plt.figure(figsize=(10,4), dpi=200)

sns.countplot(x='Survived', data=titanic)

plt.show()
```

### Observation:
 - The number of non-survivors was higher than the number of survivors.

```python
# Shows the number of male and female passengers who survived and did not survive.

plt.figure(figsize=(10,4), dpi=200)

sns.countplot(x='Sex', hue='Survived', data=titanic)

plt.xlabel('Gender')
plt.ylabel('No. of Passengers')

plt.show()
```

## Observation:
- Among male passengers, the number of non-survivors was much higher than survivors, while among female passengers, survivors were higher than non-survivors.

```python
# Shows the average survival rate of male and female passengers.

plt.figure(figsize=(10,4), dpi=200)

sns.barplot(x='Sex', y='Survived', data=titanic)

plt.xlabel('Gender')
plt.ylabel('Survival Rate')

plt.show()
```

## Observation:
- The observed survival rate was much higher for female passengers than for male passengers.

```python
# Shows the distribution of passengers ages.

plt.figure(figsize=(10,4), dpi=200)

sns.histplot(x='Age', data=titanic, bins=20)

plt.show()

```

## Observation:
- Most recorded/imputed passenger ages are concentrated around the 20–30 year range.

```python
# Shows the age distribution of passengers separately for survivors and non-survivors.

plt.figure(figsize=(10,4), dpi=200)

sns.histplot(x='Age', hue='Survived', data=titanic, bins=20)

plt.show()
```

## Observation:
- Passengers from different age groups were present in both survival groups, with non-survivors generally higher in several age ranges.

```python
# Shows the distribution of ticket fares paid by passengers.

plt.figure(figsize=(10,4), dpi=200)

sns.histplot(x='Fare', data=titanic, bins=20)

plt.show()
```

## Observation:
- Most passengers paid relatively low fares, while a small number of passengers paid very high fares.

```python
# Shows the fare distribution of passengers separately for survivors and non-survivors.

plt.figure(figsize=(10,4), dpi=200)

sns.histplot(x='Fare', hue='Survived', data=titanic, bins=20)

plt.show()
```

## Observation:
- Most passengers in both survival groups paid relatively low fares. Higher fares were less common, and in some higher-fare ranges, survivors were more than non-survivors.

```python
# Shows the average fare paid by passengers in each passenger class.

plt.figure(figsize=(10,4), dpi=200)

sns.barplot(x='Pclass', y='Fare', data=titanic)

plt.show()
```

## Observation:
- The average fare was highest for Class 1 passengers and lowest for Class 3 passengers.

```python
# Shows the distribution, median, spread, and outliers of fare across each passenger class.

plt.figure(figsize=(10,4), dpi=200)

sns.boxplot(x='Pclass', y='Fare', data=titanic)

plt.show()
```

## Observation:
- Class 1 had higher fares and more variation compared to Classes 2 and 3.

```python
# Shows the age distribution, median, spread, and outliers across each passenger class.

plt.figure(figsize=(10,4), dpi=200)

sns.boxplot(x='Pclass', y='Age', data=titanic)

plt.show()
```

## Observation:
- Class 1 passengers generally had a higher median age, while Class 3 had a younger age distribution.

```python
# Shows the survival rate of passengers across different passenger classes.

plt.figure(figsize=(10,4), dpi=200)

sns.barplot(x='Pclass', y='Survived', data=titanic)

plt.show()
```

## Observation:
- The observed survival rate was highest for Class 1 and lowest for Class 3.

```python
# Shows the average survival rate of passengers from each embarkation point.

plt.figure(figsize=(10,4), dpi=200)

e_order = titanic.groupby('Embarked')['Survived'].mean().sort_values(ascending=False).index

sns.barplot(x='Embarked', y='Survived', data=titanic, order=e_order)

plt.show()
```

## Observation:
- The observed survival rate was highest for passengers who embarked from C and lowest for those who embarked from S.

```python
# Shows the survival rate across passenger classes for male and female passengers.

plt.figure(figsize=(10,4), dpi=200)

sns.barplot(x='Pclass', y='Survived', hue='Sex', data=titanic)

plt.show()
```

## Observation
- Female passengers had a higher survival rate than male passengers across all passenger classes. Among both males and females, the survival rate was highest in Class 1 and lowest in Class 3.

# Key Insights

- Female passengers had a higher survival rate than male passengers, and survival rates also varied across passenger classes.
- Most passenger ages were around 20–30 years, and Class 1 passengers were generally older than Class 3 passengers.
- The fare distribution was right-skewed, with most passengers paying lower fares and only a few passengers paying very high fares.
